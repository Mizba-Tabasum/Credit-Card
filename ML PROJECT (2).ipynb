```python
def fix_outliers_iqr(df):
    num=df.select_dtypes(include='number').columns
    for col in num:
        Q1=df[col].quantile(0.25)
        Q3=df[col].quantile(0.75)
        IQR=  Q3 - Q1
        lower_bound=  Q1 - 1.5 * IQR
        upper_bound= Q3 + 1.5 * IQR
        df[col]=df[col].apply(lambda x: lower_bound if x < lower_bound else upper_bound if x > upper_bound else x)
    return df    
```

```python
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
skb = SelectKBest(score_func=f_classif, k=8)
```

```python
X = X_smote
y = y_smote
```

```python
X_new = skb.fit_transform(X, y)
```

```python
selected_columns = X.columns[skb.get_support()]
selected_columns
```

```python
X_selected = X[selected_columns]
X_selected
```

```python
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='yeo-johnson')

X_transformed = pd.DataFrame(
    pt.fit_transform(X_selected),
    columns=X_selected.columns
)

X_transformed
```

```python
X_transformed.skew()
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

X_scaled = ss.fit_transform(X_transformed)

X_scaled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=40
)
```

```python
print('X_train Shape:', X_train.shape)
print('X_test Shape:', X_test.shape)
print('y_train Shape:', y_train.shape)
print('y_test Shape:', y_test.shape)
```

```python
 from sklearn.linear_model import LogisticRegression
```

```python
LR= LogisticRegression()
```

```python
LR.fit(X_train, y_train)
```

```python
y_pred_lr= LR.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy= accuracy_score(y_test,y_pred_lr)
```

```python
print("accuracy:",accuracy)
```

```python
from sklearn.metrics import classification_report,f1_score,precision_score,recall_score
```

```python
accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr, average='weighted')
f1_lr = f1_score(y_test, y_pred_lr, average='weighted')
recall_lr = recall_score(y_test, y_pred_lr, average='weighted')
```

```python
print("accuracy of Logistic Regression:", accuracy_lr)
print("precision of Logistic Regression:", precision_lr)
print("f1_Score of Logistic Regression:", f1_lr)
print("recall of Logistic Regression:", recall_lr)
```

```python
print("Classification Report of Logistic Regression:\n",
      classification_report(y_test, y_pred_lr))
```
